## Background
*A Brief Recap of the Forward, Backward and Forward-Backward Algorithm*

**Viterbi**: *which* single hidden path best explains the observations.

**Forward, Backward, Forward-Backward**: To know how likely the observations are under the model at all, summed over **every** path, or how confident we should be that position *t* is in state *i* given **everything** we observed.

### Key Features
* **Joint Probability Calculation**: the Forward algorithm computes , the joint probability of the partial observation sequence and ending in state *i*
* **Conditional Future Probability**: the Backward algorithm computes , the probability of future observations given the current state.
* **Posterior Marginals**: the Forward-Backward algorithm combines these to determine , the probability of being in each state at each position given the entire observation sequence.
* **Dynamic Programming**: efficient recursion avoids exponential complexity

### Underflow
*A sum does not pass straight through a logarithm*

1. *scaling*: normalizes each column of the forward (and backward) matrix as it is computed and keeps track of the scale factors.

2. *log-sum-exp* trick computes  stably factoring out the largest term
    * software resolution: work with logarithms throughout, so that products become sums
